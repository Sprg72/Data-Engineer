````python
# way1
gcnt = {}
gtot = {}
gmax = {}
gmin = {}

for x in emp:
  gend = x[-2]
  sal = x[2]
  if gcnt.get(gend) == None:
    gcnt[gend] = 1
    gtot[gend] = sal
    gmax[gend] = sal
    gmin[gend] = sal
  else:
    gcnt[gend] += 1
    gtot[gend] += sal
    gmax[gend] = max(gmax[gend], sal)
    gmin[gend] = min(gmin[gend], sal)

gavg = {}
for g in gcnt:
  cnt = gcnt[g]
  tot = gtot[g]
  avg = round(tot/cnt,2)
  gavg[g] = avg


print(f"count : {gcnt}")
print(f"total : {gtot}")
print(f"average :  {gavg}")
print(f"max   : {gmax}")
print(f"min   : {gmin}")
````

````python
# ETL ----> extract, transform and load

# expected transformations
# id ------------> id
# name ----------> first letter upppercase , remaining letters into lower case
# salary --------> salary
# tax -----------> 10% on salary
# hra -----------> 20% on salary
# netsalary -----> salary + hra - tax
# Grade ---------> based on salary ranges.
#           if salary >=100000             -----> Grade A
#           if salary >=75000 and <100000  -----> Grade B
#           if salary >=50000 and <750000  -----> Grade C
#           if salary < 50000              -----> Grade D
# Gender ------->
#              'm' ------> 'Male'
#              'f' ------> 'Female'
#              remaining-> 'Invalid'
# dno ----------> dno
# dname --------> based on dnos
#            11 ---> Marketing
#            12 ---> Hr
#            13 ---> Finance
#           remainig -> other


def firstUpper(s):
  s = s.strip().lower()
  fc = s[0].upper()
  rc = s[1:]
  return fc + rc

def toGrade(sal):
  grade = 'D'
  if sal >=100000:
    grade = 'A'
  elif sal >= 75000:
    grade = 'B'
  elif sal >= 50000:
    grade = 'C'
  return grade


def toGender(g):
  g = g.strip().lower()
  if g == 'm':
     gend = 'Male'
  elif g == 'f':
     gend = 'Female'
  else:
     gend = 'Invalid'
  return gend


def toDept(dno):
    dinfo = {11: 'Marketing', 12: 'Hr', 13: 'Finance'}
    dname = dinfo.get(dno, 'Other')
    return dname
# "101, amar,90000,m,11\n"
def transform(line):
  line = line.strip().lower()
  w = line.split(',')
  print(w)
  #['101', ' amar', '90000', 'm', '11']
  id = w[0]
  name = firstUpper(w[1])
  sal = int(w[2])
  tax = round(sal * 0.1, 2)
  hra = round(sal * 0.2, 2)
  netsal = round(sal + hra - tax, 2)
  grade = toGrade(sal)
  gender = toGender(w[-2])
  dno = int(w[-1])
  dname = toDept(dno)
  res = [id, name, sal, tax, hra, netsal, grade, gender, dno, dname]
  res = [str(v) for v in res]
  return res








````

````python
# "101, amar,90000,m,11\n"
def transform(line):
  line = line.strip().lower()
  w = line.split(',')

  #['101', ' amar', '90000', 'm', '11']
  id = w[0]
  name = firstUpper(w[1])
  sal = int(w[2])
  tax = round(sal * 0.1, 2)
  hra = round(sal * 0.2, 2)
  netsal = round(sal + hra - tax, 2)
  grade = toGrade(sal)
  gender = toGender(w[-2])
  dno = int(w[-1])
  dname = toDept(dno)
  res = [id, name, sal, tax, hra, netsal, grade, gender, dno, dname]
  res = [str(v) for v in res]
  resline = ','.join(res) + '\n'
  return resline
````

Spark data objects called RDD's ( resilient distributed datasets)

what is RDD --> is a logical name and is collection of data partitions.
                Runtime these partitions are loaded in RAM's of different nodes of the cluster
                These partitions will be executed by multi node paralley(MPP)

Two types of operations on RDD are transformations and actions.
RDD transformations:
eg. RDD.map()
    RDD.flatMap()
    RDD.filter()
    RDD.groupBykey()
    RDD.reduceByKey()

RDD actions:
eg. RDD.collect()
    RDD.saveAsTextFile()
    RDD.sum()
    RDD.reduce()

When you perform transformation in existing RDD, A new RDD will be declared. But won't be executed.

Whe you perform action on RDD, entire dataflow will be executed from its root RDD

RDD Case1
x=[10,20,30,40,50,60]

r1 = sc.parallelize(x, 2)             #   root         #r1 declared
r2 = r1.map(lambda v: v+100)          # transformation #r2 decalred
r3 = r2.filter(lambda x : x>130)      # transformation #r3 declared
r3.collect() # action

---> the script will subitted to 'DAG' engine. (DAG -- Directed Acyclic Graph)

Dataflow : R1 --> R2 --> R3

The flow will be executed from R1

This style of execution is called "Lazy Evolution"
